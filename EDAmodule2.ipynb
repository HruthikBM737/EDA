```python
import pandas as pd
data = [45, 67, 89, 23, 56]
s = pd.Series(data)
print("Series:")
print(s)
print("\nValues:")
print(s.values)
print("\nIndex:")
print(s.index)
```

```python
import pandas as pd
population_data = [20_411_000, 18_980_000, 12_327_000, 11_503_000, 14_850_000]
cities = ['Mumbai', 'Delhi', 'Bengaluru', 'Chennai', 'Kolkata']
population = pd.Series(population_data, index=cities)
print("Population Series:")
print(population)
print("\nPopulation of Bengaluru:")
print(population['Bengaluru'])
print("\nCities from Delhi to Chennai:")
print(population['Delhi':'Chennai'])
```
