```python
import pandas as pd
data = {
    'department': ['CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE',
                   'CSE', 'AIML', 'ECE', 'CSE', 'AIML', 'ECE'],
    'gender': ['M', 'F', 'M', 'F', 'M', 'F',
               'M', 'F', 'M', 'F', 'M', 'F'],
    'salary': [50000, 55000, 48000, 60000, 53000, 52000,
               58000, 62000, 51000, 61000, 57000, 54000]
}
df = pd.DataFrame(data)
print("Employee Data:")
print(df)
pivot_table = pd.pivot_table(
    df,
    values='salary',
    index='gender',
    columns='department',
    aggfunc='mean'
)
print("\nAverage Salary by Gender and Department:")
print(pivot_table)
```
